# Modelica ISO 13790 BESTEST executor

Run annual reference simulations from the Modelica Buildings Library.

## Runtime and path configuration

The setup cell below locates `_modelica` when Jupyter is launched from the repository root, `_modelica`, or one of their subdirectories. The Modelica Buildings checkout is discovered from the usual repository locations. On another computer or in another layout, set `MODELICA_BUILDINGS_ROOT` to the directory containing `Buildings/package.mo`.

1. **Native OpenModelica (validated):** leave the configuration unchanged when `omc` is on `PATH`. For a nonstandard installation, uncomment `OPENMODELICA_OMC` and enter its absolute path.
2. **Docker (alternative, untested):** use this only when native `omc` is unavailable. Leave the configuration unchanged when `docker` is on `PATH`; otherwise uncomment `DOCKER_BIN` and enter its absolute path. The launcher prefers native `omc` when both runtimes are available.


```python
from pathlib import Path
import os
import sys

cwd = Path.cwd().resolve()
candidates = [candidate for base in (cwd, *cwd.parents) for candidate in (base, base / "_modelica")]
MODELICA_ROOT = next((path for path in candidates if (path / "_validation/modelica_runner.py").is_file()), None)
if MODELICA_ROOT is None:
    raise FileNotFoundError("Could not locate _modelica/_validation/modelica_runner.py")
sys.path.insert(0, str(MODELICA_ROOT))

# Native OpenModelica at a nonstandard location:
# os.environ["OPENMODELICA_OMC"] = "/absolute/path/to/omc"

# Docker CLI at a nonstandard location (alternative, untested):
# os.environ["DOCKER_BIN"] = "/absolute/path/to/docker"

# Modelica Buildings checkout in a nonstandard location (either runtime):
# os.environ["MODELICA_BUILDINGS_ROOT"] = "/absolute/path/to/modelica_test"

from _validation.modelica_runner import run_modelica_case
BESTEST_BASE = "Buildings.ThermalZones.ISO13790.Validation.BESTEST"
CASES = {
    "BESTEST-600FF": "Cases6xx.Case600FF",
    "BESTEST-900FF": "Cases9xx.Case900FF",
}
```

```python
modelica_results = {}
for case, model_path in CASES.items():
    print(f"Running {case}...")
    model_class = f"{BESTEST_BASE}.{model_path}"
    modelica_results[case] = run_modelica_case(case, model_class)

result_shapes = {
    case: frame.shape
    for case, frame in modelica_results.items()
}
result_shapes
```
