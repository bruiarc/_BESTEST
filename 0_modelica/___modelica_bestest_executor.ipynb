# Modelica ISO 13790 BESTEST reference runs

Runs the five BESTEST cases using only the pinned local `modelica-buildings-github-master` copy. Each CSV is a secondary Modelica implementation reference; BESTEST acceptance remains a separate comparison.

```python
from pathlib import Path
import sys

CWD = Path.cwd().resolve()
ROOT = next(p for b in (CWD, *CWD.parents) for p in (b, b / '0_modelica') if (p / 'iso_validation').is_dir())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from iso_validation.modelica_runner import run_modelica_class as _run_modelica_class

BUILDINGS_ROOT = ROOT / 'modelica-buildings-github-master'
assert (BUILDINGS_ROOT / 'Buildings' / 'package.mo').is_file(), BUILDINGS_ROOT

CASES = {
    '600FF': 'Buildings.ThermalZones.ISO13790.Validation.BESTEST.Cases6xx.Case600FF',
    '900FF': 'Buildings.ThermalZones.ISO13790.Validation.BESTEST.Cases9xx.Case900FF',
    '600': 'Buildings.ThermalZones.ISO13790.Validation.BESTEST.Cases6xx.Case600',
    '900': 'Buildings.ThermalZones.ISO13790.Validation.BESTEST.Cases9xx.Case900',
    '950': 'Buildings.ThermalZones.ISO13790.Validation.BESTEST.Cases9xx.Case950',
}

def run_modelica_case(case: str):
    model_class = CASES[case]
    df, source, csv_path = _run_modelica_class(
        ROOT, model_class, buildings_root=BUILDINGS_ROOT
    )
    print(f'{case}: {model_class}')
    print('Source:', source)
    print('CSV:', csv_path)
    print('Time range:', df.time.iloc[0], 'to', df.time.iloc[-1])
    print('Rows/columns:', df.shape)
    return df
```

```python
# Annual BESTEST simulations. Run once; individual CSVs are retained in results/.
modelica_results = {case: run_modelica_case(case) for case in CASES}
{case: frame.shape for case, frame in modelica_results.items()}
```
