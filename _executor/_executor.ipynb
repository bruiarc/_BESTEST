# Modelica ISO 13790 BESTEST executor

Run annual reference simulations from the Modelica Buildings Library.

```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
candidates = [candidate for base in (cwd, *cwd.parents) for candidate in (base, base / "_executor")]
EXECUTOR_ROOT = next((path for path in candidates if (path / "core/modelica_runner.py").is_file()), None)
if EXECUTOR_ROOT is None:
    raise FileNotFoundError("Could not locate _executor/core/modelica_runner.py")
sys.path.insert(0, str(EXECUTOR_ROOT))

from core.modelica_runner import run_modelica_case
BESTEST_BASE = "Buildings.ThermalZones.ISO13790.Validation.BESTEST"
CASES = {
    "BESTEST-600FF": "Cases6xx.Case600FF",
    "BESTEST-900FF": "Cases9xx.Case900FF",
}
```

```python
modelica_results = {}
for case, model_path in CASES.items():
    print(f"Running {case}...")
    model_class = f"{BESTEST_BASE}.{model_path}"
    modelica_results[case] = run_modelica_case(case, model_class)

result_shapes = {
    case: frame.shape
    for case, frame in modelica_results.items()
}
result_shapes
```
